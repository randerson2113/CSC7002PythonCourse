### POS Tagging


```python
# POS tagging - return counts

# tag the parts from short_blob
short_tags = short_blob.tags

# for tag in short_tags:
#     print(tag)

# count the tag types
tag_counts = {}
for word, tag in short_blob.tags:
    if tag not in tag_counts:
        tag_counts[tag] = 1
    else:
        tag_counts[tag] += 1

# sort tag_counts by count in descending order
sorted_tag_counts = sorted(tag_counts.items(), key=lambda item: item[1], reverse=True)

print("\nPOS Tag Counts (Sorted by most common) and Descriptions:")
# print the word counts for each tag type and their descriptions
for tag, count in sorted_tag_counts:
    description = pos_tags_description.get(tag, 'Description not available')
    print(f"{tag}: {count} ({description})")

# list all distinct words of a particular type
nouns = [word for word, tag in short_blob.tags if tag == 'NN']
print("\nNouns:")
print(nouns)
```

### Inter-language translation


```python
from textblob import TextBlob
from googletrans import Translator # import the Google translator

# create multilingual text for translation
multilingual_text = """TextBlob is an amazing library for NLP tasks! ¡Es muy fácil de usar y muy potente! J'aime beaucoup la simplicité de cet outil pour le traitement de texte. Das ist wirklich ein sehr hilfreiches Werkzeug für meine Arbeit. Hopefully, it can handle this mix."""

translator = Translator() # Instantiate Translator

# translate to all the languages above, by sentence

blob_english = TextBlob(translator.translate(multilingual_text, dest='en').text)
blob_espanol = TextBlob(translator.translate(multilingual_text, dest='es').text)
blob_francais = TextBlob(translator.translate(multilingual_text, dest='fr').text)
blob_deutsch = TextBlob(translator.translate(multilingual_text, dest='de').text)

# print each sentence of the translated texts for readability
for blob in [blob_english, blob_espanol, blob_francais, blob_deutsch]:
  print(blob)




```

### Inflection


```python
# inflection demo - singularization and pluralization

blob_english_inflected = blob_english.words
print(blob_english_inflected)

blob_english_pluralized = blob_english_inflected.pluralize()
print(blob_english_pluralized)

blob_english_singularized = blob_english_inflected.singularize()
print(blob_english_singularized)

# skip words that already end in s
for word in blob_english_inflected:
  if word.endswith('s') or word.endswith('es') or word.endswith('ies'):
    print(word.singularize())
  elif not(word.endswith('s') or word.endswith('es') or word.endswith('ies')):
    print(word.pluralize())
  else:
    print(word)
```


### Stemming


```python
# stemming words
stemmed_words = [word.stem() for word in blob_english_singularized]

for i in range(len(blob_english_singularized)):
  print(f"{blob_english_singularized[i]} -> {stemmed_words[i]}")
```

### Lemmatization


```python
# lemmatization demo
lemmatized_words = [word.lemmatize() for word in blob_english_singularized]

print(lemmatized_words)
```

### WordNet Integration


```python
# wordnet demo
from textblob import Word
from nltk.corpus import wordnet as wn

# wordnet functions demo on a blob
demo_word = blob_english_singularized[23] # This is a TextBlob.Word object, e.g., Word('tool')

# Get all synsets for the word using NLTK's wordnet
# wn.synset() expects a string like 'lemma.pos.num', so we use wn.synsets() for a word
all_synsets = wn.synsets(str(demo_word))

if all_synsets: # Check if any synsets are found
    # For demonstration, let's work with the first synset
    new_word = all_synsets[0]

    print(f"Word: {demo_word}")
    print(f"Selected Synset: {new_word.name()}")

    # Get synonyms
    synonyms = sorted(new_word.lemma_names())
    print(f"Synonyms: {synonyms}")

    # Get hypernyms
    hypernyms = sorted([s.name() for s in new_word.hypernyms()])
    print(f"Hypernyms: {hypernyms}")

    # Get hyponyms
    hyponyms = sorted([s.name() for s in new_word.hyponyms()])
    print(f"Hyponyms: {hyponyms}")
else:
    print(f"No synsets found for '{demo_word}'")
```
